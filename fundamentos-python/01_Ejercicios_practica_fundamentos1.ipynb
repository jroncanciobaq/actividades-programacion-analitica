```python
#Sin ejecutar, determina el valor final de suma.
suma = 0

for i in range (1,4):
  suma = suma + i

  print(suma)

#El valor final de la suma daría 1, 3 y 6 ya que inicialmente como suma = 0, le adiciona el valor de i que es uno y muestra 1;
#luego como suma es 1, le adiciona el segundo valor de i que es dos y muestra 3
#finalmente como suma es 3, le adiciona el valor de i que es 3 y muestra 6
#No muestra 10 porque la función Range solo muestra el valor superior - 1.
#es decir que range solo muestra valores de 1,2 y 3, los reemplaza en i y cada valor de i se los suma a la variable "suma" y el ciclo empieza de nuevo hasta que el valor de i = 3
```

### G. Del psudocodigo a Python
